```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/krishnaik06/playstore-Dataset/main/googleplaystore.csv")
df.head()
```

```python
df.shape
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Reviews"].str.isnumeric().sum()
```

```python
#it will give that coloumn that is not numeric as total values are 10841 and is numeric is 10840
df[~df["Reviews"].str.isnumeric()]
```

```python
df.head(2)
```

```python
df_copy = df.copy()
df_copy=df_copy.drop(df_copy.index[10472])
```

```python
df_copy["Reviews"] = df_copy["Reviews"].astype(int)
df_copy.info()
```

```python
df_copy["Size"].unique()
```

```python
df_copy["Size"] = df_copy["Size"].replace("Varies with device", np.nan)
df_copy["Size"] = df_copy["Size"].astype(str)

df_copy["Size"] = df_copy["Size"].str.replace("M", "000")
df_copy["Size"] = df_copy["Size"].str.replace("k", "")

df_copy["Size"] = df_copy["Size"].astype(float)

df_copy.info()
```

```python
df_copy["Price"].unique()
```

```python
df_copy["Installs"].unique()
```

```python
char_to_remove = ["+",",","$"]
cols_to_clean = ["Price","Installs"]

for cols in cols_to_clean:

  df_copy[cols] = df_copy[cols].astype(str)
  for i in char_to_remove:
    df_copy[cols] = df_copy[cols].str.replace(i,"")
```

```python
df_copy["Price"].unique()
```

```python
df_copy["Installs"].unique()
```

```python
df_copy["Installs"] = df_copy["Installs"].astype(float)
```

```python
df_copy["Price"] = df_copy["Price"].astype(float)
```

```python
df_copy.info()
```

```python
df_copy["Last Updated"] = pd.to_datetime(df_copy["Last Updated"])
df_copy["Last Updated_date"] = df_copy["Last Updated"].dt.day
df_copy["Last Updated_month"] = df_copy["Last Updated"].dt.month
df_copy["Last Updated_year"] = df_copy["Last Updated"].dt.year
df_copy.info()
```

```python
df_copy = df_copy.drop("Last Updated", axis = 1)
```

```python
df_copy.info()
```

```python
df_copy.to_csv("/google_cleaned_csv")
```

# **Eploring data **

```python
numeric_feature = [feature for feature in df_copy.columns if df_copy[feature].dtype != 'object']
categorical_feature = [feature for feature in df_copy.columns if df_copy[feature].dtype == 'object']
```

```python
print("we have {} numerica_feature:{}".format(len(numeric_feature),numeric_feature))
print("we have {} categorical_feature:{}".format(len(categorical_feature),categorical_feature))
```

```python
df_copy.head(2)
```

```python
for cols in categorical_feature:
  print(df_copy[cols].value_counts(normalize = True)*100)
  print("------------------------------")
```

```python
plt.figure(figsize = (15,15))
plt.suptitle("univarite Analysis of numeric_feature",fontsize = 20,fontweight = "bold",alpha = 0.8,y = 1)
for i in range (0,len(numeric_feature)):
  plt.subplot(4,4,i+1)
  sns.kdeplot(x = df_copy[numeric_feature[i]], fill = True, color = "red")
  plt.xlabel(numeric_feature[i])
plt.tight_layout()
plt.show()
```
